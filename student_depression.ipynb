```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
filepath = '/content/drive/MyDrive/Colab Notebooks/dataset/Student_Depression_Dataset.csv'
```

```python
stu_df= pd.read_csv(filepath)
stu_df.head(10)
```

```python
stu_df.info()
```

```python
stu_df.describe()
```

```python
#creating a cloumn storing all the numerical column names
num_df = stu_df.select_dtypes(include=['number'])
##creating a cloumn storing all the categorical column names
cat_df = stu_df.select_dtypes(include=['object'])

```

```python
num_df
```

```python
num_df.isna().sum()
```

```python
stu_df= stu_df.drop_duplicates()
stu_df.shape
```

```python
cat_df

```

```python
cat_df.isna().sum()
```

```python
stu_df['Depression'].unique()
```

```python
label_encoder = LabelEncoder()
stu_df['Depression'] = label_encoder.fit_transform(stu_df['Depression'])
```

```python
stu_df['Depression'].unique()
```

```python
stu_df.head(10)
```

```python
stu_df['Health_condition'].unique()
```

```python
encoder_health = OrdinalEncoder(
    categories=[['Critical', 'Abnormal', 'Good', 'Normal']]
)
stu_df['Health_condition'] = encoder_health.fit_transform(stu_df[['Health_condition']])
```

```python
stu_df.head(10)
```

```python
stu_df['Family_background'].unique()
```

```python
encoder_family = OrdinalEncoder(
    categories=[['Lower Middle class', 'Middle class', 'Rich class']]
)
stu_df['Family_background'] = encoder_family.fit_transform(stu_df[['Family_background']])
```

```python
stu_df.head(10)
```

```python
encoder = OneHotEncoder()
encoder_data = encoder.fit_transform(cat_df[["Gender"]])

encoder_stu_df = pd.DataFrame(
    encoder_data.toarray(),
    columns=encoder.get_feature_names_out(['Gender']),
    index=cat_df.index
)
stu_df = pd.concat([stu_df, encoder_stu_df],axis=1).drop(columns=['Gender'])
```
